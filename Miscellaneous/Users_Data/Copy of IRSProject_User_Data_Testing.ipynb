```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

model_name = "Qwen/Qwen2.5-3B-Instruct"

tokenizer = AutoTokenizer.from_pretrained(model_name)

model = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype=torch.float16,
    device_map="auto"
)
```

```python
def build_prompt(profile_text):

    return f"""
You are extracting user interests for an activity recommender system.

From the following profile, extract only interests that could help recommend:
- activities
- restaurants
- cafes
- events
- travel experiences
- entertainment
- outdoor activities
- cultural places

Return valid JSON only.

Use this structure:

{{
  "hobbies": [],
  "music_genres": [],
  "music_vibes": [],
  "food_interests": [],
  "sports": [],
  "outdoor_activities": [],
  "indoor_activities": [],
  "arts_culture": [],
  "nightlife": [],
  "travel_interests": [],
  "activity_vibes": []
}}

Rules:
- Do not invent interests.
- Extract only what is stated or strongly implied.
- Ignore dating preferences.
- Ignore sex, orientation, religion, ethnicity, and appearance.
- Normalize similar concepts.
- Use short phrases.

PROFILE:
{profile_text}
"""
```

```python
def extract_interests(profile_text):

    prompt = build_prompt(profile_text)

    messages = [
        {"role": "user", "content": prompt}
    ]

    text = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(
        text,
        return_tensors="pt"
    ).to(model.device)

    outputs = model.generate(
        **inputs,
        max_new_tokens=400,
        temperature=0.1,
        do_sample=False
    )

    result = tokenizer.decode(
        outputs[0][inputs.input_ids.shape[1]:],
        skip_special_tokens=True
    )

    return result
```

```python
result = extract_interests(df.loc[0, "profile_text"])

print(result)
```
